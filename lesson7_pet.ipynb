## 1.2. Первичный осмотр
Выведите: размер таблицы, типы данных, первые 5 строк, описательную статистику.

**Объяснение:** Напишите текстом, что вы видите: какие типы неправильные, есть ли пропуски, есть ли странные значения (например, отрицательный возраст, дата в будущем).

```python
df = pd.read_csv('ecommerce_retail_transactions_raw.csv')
df.info()
df.describe()
```

в колонках 'Discount_Percent', 'Shipping_City', 'Customer_Rating' есть пустые значения, в колонке 'Order_Date' строковый тип данных, что не позволяет работать с датами, данные в колонках 'Payment_Method', 'Shipping_City', 'Country' заполняют люди, поэтому там могут быть дубликаты. в колонке 'Quantity" минимальное -1, но отрицательного количества товаров не может быть, значит где-то сбой

## 1.3. Дубликаты
Найдите количество полных дубликатов и удалите их.

**Объяснение:** Сколько дублей было? По какому принципу удаляли?

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

полных дубликатов было 180, удалили  потому что полные дубликаты искажают информацию, то есть при подсчете среднего чека например , одна транкзация посчиталась несколько раз, а это означает что данные будут неверные

## 1.4. Пропуски
Посчитайте пропуски по колонкам. Решите, что с ними делать: удалить строки или заполнить.

**Объяснение:** Почему выбрали именно такой метод? 

```python
df.isnull().sum()
df['Discount_Percent'] = df['Discount_Percent'].fillna(0)
df['Shipping_City'] = df['Shipping_City'].fillna('Unknown')
df['Customer_Rating'] = df['Customer_Rating'].fillna(df['Customer_Rating'].median())
```

колонку 'Discount_Percent' заполнили 0, потому что 0 означает, что скидки не было. колонку 'Shipping_City' заполнили надписью 'Unknown', потому что иначе не возможно проанализировать географию заказов. колонку 'Customer_Rating' заполнили медианой, потому что будет +- объективный рейтинг

## 1.5. Типы данных и текст
Приведите типы к правильным: даты в `pd.to_datetime`, числа из текста в `int`/`float`. Уберите лишние пробелы в тексте и приведите регистр к нижнему.

**Объяснение:** Какие колонки были неправильного типа? Как чистили текст?

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed')
df['Order_ID'] = df['Order_ID'].str.lower().str.strip()
df['Customer_ID'] = df['Customer_ID'].str.lower().str.strip()
df['Product_Category'] = df['Product_Category'].str.lower().str.strip()
df['Product_Name'] = df['Product_Name'].str.lower().str.strip()
df['Payment_Method'] = df['Payment_Method'].str.lower().str.strip()
df['Shipping_City'] = df['Shipping_City'].str.lower().str.strip()
df['Country'] = df['Country'].str.lower().str.strip()
df['Order_Status'] = df['Order_Status'].str.lower().str.strip()

```

в колонке 'Order_Date' даты были записаны str типом, поменяли на to_datetime применим format='mixed', потому что даты были разных форматов, в остальных строках с str типов привели все к нижнему регистру и убрали лишние пробелы
